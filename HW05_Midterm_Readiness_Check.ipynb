```python
# Question 1 analysis and visualization

# YOUR CODE HERE
import matplotlib.pyplot as plt

songs = df.drop_duplicates("track_id")
by_genre = df.drop_duplicates(["track_id", "playlist_genre"])

genre_mood = by_genre.groupby("playlist_genre")[["valence", "energy"]].mean().sort_values("valence")
print(genre_mood.round(3))

plt.figure(figsize=(8, 6))
plt.scatter(genre_mood["valence"], genre_mood["energy"], s=200)
for name, row in genre_mood.iterrows():
    plt.text(row["valence"], row["energy"] + 0.012, name, ha="center")
plt.axhline(0.5, color="gray", ls="--")
plt.axvline(0.5, color="gray", ls="--")
plt.xlim(0, 1)
plt.ylim(0, 1)
plt.xlabel("valence (sad to happy)")
plt.ylabel("energy (calm to intense)")
plt.title("Where each genre sits on the mood map")
plt.show()
```

**Question 1 interpretation and limitation:** This looks at whether the siz genres actually have any different moods based on valence and energy. The averages show us that they do sit in different spots, edm being the highest energy at 0.80 but the lowest valence at 0.41 so it feels intense but not very positive while latin has the highest valence at 0.60 and r&b has the lowest energy at 0.59. This supports my app idea because it means that by filtering genre it also changes the kind of mood you get. One limitation is that genre here comes from the playlist a song was already on not an official label and these are just averages so individual songs in a genre still spread all over the mood map.

#### Concept A

**Chosen concept:** Transformation / Feature Engineering

```python
# Concept A analysis

# YOUR CODE HERE
songs = df.drop_duplicates("track_id").copy()
by_genre = df.drop_duplicates(["track_id", "playlist_genre"]).copy()

def get_mood(row):
    if row["valence"] >= 0.5 and row["energy"] >= 0.5:
        return "Positive + energetic"
    elif row["valence"] >= 0.5:
        return "Positive + calm"
    elif row["energy"] >= 0.5:
        return "Negative + energetic"
    else:
        return "Negative + calm"

songs["mood"] = songs.apply(get_mood, axis=1)
by_genre["mood"] = by_genre.apply(get_mood, axis=1)

moods = ["Positive + energetic", "Positive + calm",
         "Negative + energetic", "Negative + calm"]

mood_percent = songs["mood"].value_counts(normalize=True) * 100
print(mood_percent.reindex(moods).round(1))
```

**Concept A interpretation / decision:** Basically, I used feature engineering to create four mood groups using valence and energy, which have a 0.5 as the cutoff. This is useful because the dataset doesn't actually have any mood labels. I found that 46.6% of songs are positive and energetic, and 38.1% are negative and energetic. So most songs have high energy regardless of their valence. These aren't actual listener mood ratings, just categories I made from the audio features. I'm going to use them in my Streamlit app so users can choose a mood and then find matching songs as well.

#### Concept B

**Chosen concept:**  Distribution / Histogram / KDE

```python
# Concept B analysis

# YOUR CODE HERE
import matplotlib.pyplot as plt

genre_moods = pd.crosstab(by_genre["playlist_genre"], by_genre["mood"], normalize="index") * 100

print(genre_moods[moods].round(1))

genre_moods[moods].plot(kind="bar", stacked=True, figsize=(10, 5))
plt.title("Mood groups within each genre")
plt.xlabel("Genre")
plt.ylabel("Songs (%)")
plt.xticks(rotation=0)
plt.legend(title="Mood", bbox_to_anchor=(1, 1))
plt.tight_layout()
plt.show()
```

**Concept B interpretation / decision:** I wanted to look at the distribution of moods within each genre instead of only comparing averages. So I used percentages because the genres don't all have the same number of songs to begin with. Around 63.6% of EDM songs are negative and energetic while 63% of Latin songs are positive and energetic. This shows us that the genres have a pretty different mood distribution, even though songs within the same genre can still vary a lot. This supports adding genre and mood filters to my app so that users can explore those differences themselves as well.
